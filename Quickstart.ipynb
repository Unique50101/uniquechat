```python
from uniquechat.engine import Engine, TEXT, read_turn

model = Engine(TEXT, cached=False)
history = []

def ask(question, files=()):
    turn = read_turn(question, files, model.vision)
    reply = model.reply(history + [turn])
    history.extend([turn, {"role": "assistant", "content": [{"type": "text", "text": reply}]}])
    return reply

print(ask("What is the capital of France? Answer briefly."))
```

```python
print(ask("Which country is that city in?"))
```
